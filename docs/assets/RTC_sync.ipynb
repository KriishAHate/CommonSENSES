```python
import serial
import time
from datetime import datetime

# Specify the serial port your Arduino is connected to (change 'COM3' as needed)
arduino = serial.Serial('COM5', 9600)  # Windows: 'COM3', Linux: '/dev/ttyACM0' or '/dev/ttyUSB0'
time.sleep(2)  # Wait for Arduino to reset

while True:
    # Get the current time from the computer
    current_time = datetime.now().strftime("%Y-%m-%d %H:%M:%S")
    
    # Send the time to Arduino
    arduino.write(current_time.encode() + b'\n')
    
    # Wait a second before sending again (you can change this interval)
    print(f"Sent time: {current_time}")
    time.sleep(1)
```
